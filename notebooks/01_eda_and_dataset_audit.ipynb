# SafeGuard — Notebook 01: Dataset EDA, Harmonization & Leakage Audit

This notebook provides an interactive exploratory entry point calling the modular, tested scripts in `src/data/` and summarizing `reports/dataset/dataset_audit.md` and `reports/dataset/final_split_statistics.json`.
**Note**: Canonical production pipelines run via `dvc.yaml` and `src/orchestration/prefect_flows.py` (never notebook-only state).

```python
import json
from pathlib import Path
import yaml

ROOT = Path('..') if Path.cwd().name == 'notebooks' else Path('.')
data_cfg = yaml.safe_load((ROOT / 'configs' / 'data.yaml').read_text(encoding='utf-8'))
class_map = yaml.safe_load((ROOT / 'configs' / 'class_mapping.yaml').read_text(encoding='utf-8'))

print('Pinned sources:', list(data_cfg['sources'].keys()))
print('Approved classes:', [c['name'] for c in class_map['canonical_classes'] if c.get('status') == 'approved'])
```

```python
stats_path = ROOT / 'reports' / 'dataset' / 'final_split_statistics.json'
if stats_path.exists():
    stats = json.loads(stats_path.read_text(encoding='utf-8'))
    print('Leakage checks:', json.dumps(stats['leakage_check'], indent=2))
    for part, info in stats['partitions'].items():
        print(f"{part:20s}: {info['images']:4d} images | {info['total_boxes']:5d} boxes")
else:
    print('Run `dvc repro` or `python -m src.orchestration.prefect_flows` to populate dataset statistics.')
```
