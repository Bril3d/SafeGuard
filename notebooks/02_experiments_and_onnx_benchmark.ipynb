# SafeGuard — Notebook 02: Experiment Matrix (`EXP-A`–`EXP-F`), ONNX Parity & Drift Proxy Audit

Interactive inspection notebook reviewing logged MLflow runs, ONNX verification/benchmark outputs (`reports/benchmark/`), and proxy drift inspection via `src/monitoring/drift.py`.

```python
import json
from pathlib import Path

ROOT = Path('..') if Path.cwd().name == 'notebooks' else Path('.')
verif_path = ROOT / 'reports' / 'benchmark' / 'verification_report.json'
bench_path = ROOT / 'reports' / 'benchmark' / 'benchmark_report.json'

if verif_path.exists():
    verif = json.loads(verif_path.read_text(encoding='utf-8'))
    print('ONNX same-tensor match rate:', verif['same_tensor_prediction_parity']['match_rate_vs_pytorch'])
    print('ONNX E2E match rate:', verif['prediction_parity']['match_rate_vs_pytorch'])

if bench_path.exists():
    bench = json.loads(bench_path.read_text(encoding='utf-8'))
    for rt in bench['runtimes']:
        print(f"{rt['runtime']:42s} | mean E2E={rt['end_to_end_latency']['mean_ms']:.2f}ms | FPS={rt['end_to_end_latency']['fps']:.2f}")
```
